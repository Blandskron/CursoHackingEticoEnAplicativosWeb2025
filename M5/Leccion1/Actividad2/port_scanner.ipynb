# Documentación Técnica – `scan_ports.py`


## Objetivo del Script

Este script realiza un **escaneo de puertos TCP** en un rango específico (por defecto del 20 al 1024) sobre un host determinado. Su propósito es detectar qué puertos están **abiertos** y escuchando conexiones, una fase crítica del reconocimiento activo en auditorías de seguridad.

Es funcionalmente equivalente a un **port scanner básico**, emulando el comportamiento de herramientas como `Nmap` en modo TCP Connect Scan (`-sT`), pero escrito desde cero usando `socket`.


## Librerías Utilizadas

```python
import socket
```

* `socket`: módulo estándar de Python que permite la comunicación de red. Se usa para establecer conexiones TCP manuales al host y puerto especificado.

## Función principal

```python
def scan_ports(host, start_port=20, end_port=1024):
```

* La función recibe un **host** (IP o nombre de dominio), y un rango de puertos definido por `start_port` y `end_port`.
* El rango predeterminado va del 20 al 1024, que incluye puertos conocidos y comúnmente utilizados por servicios estándar.


### Proceso de escaneo

```python
for port in range(start_port, end_port + 1):
```

* Se itera sobre cada puerto dentro del rango definido.

```python
s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
```

* Se crea un **socket TCP IPv4** para cada puerto. El protocolo utilizado es **AF\_INET** (IPv4) y **SOCK\_STREAM** (TCP).

```python
s.settimeout(0.5)
```

* Se establece un timeout corto (0.5 segundos) para evitar que el script se bloquee esperando puertos cerrados o filtrados.

```python
result = s.connect_ex((host, port))
```

* `connect_ex()` intenta realizar una conexión TCP al host y puerto actual.
* Devuelve `0` si la conexión fue exitosa (es decir, el puerto está **abierto**).
* Cualquier otro valor indica que el puerto está **cerrado**, **filtrado**, o **inaccesible**.

```python
if result == 0:
    print(f"Puerto {port}: ABIERTO")
```

* Se imprime en consola únicamente si el puerto respondió a la conexión, lo cual indica un servicio activo en ese puerto.

```python
s.close()
```

* El socket se cierra explícitamente para liberar recursos del sistema.


### Manejo de errores

```python
except socket.gaierror:
    print("Error: El host no fue encontrado.")
```

* Ocurre si el nombre de dominio no puede resolverse a una IP válida.

```python
except socket.error:
    print("Error: No se pudo establecer conexión.")
```

* Captura errores generales de red, como caídas de conectividad, tiempo de espera agotado, o bloqueos.

## Relevancia en Ciberseguridad

Este script es una herramienta de **reconocimiento activo**, clave en auditorías y pentests. Permite detectar servicios expuestos, evaluar la superficie de ataque y reconocer la estructura de red de un sistema.

La apertura de un puerto sugiere la existencia de un servicio que podría tener vulnerabilidades, especialmente si no está protegido o actualizado. Un escaneo rápido de puertos como este, aunque limitado, permite iniciar el mapeo de activos en una infraestructura objetivo.


## Consideraciones adicionales

* Se recomienda ejecutarlo solo en entornos controlados o con permiso explícito.
* Puede ampliarse con:

  * Escaneo UDP (requiere cambios profundos)
  * Identificación de servicios (banner grabbing)
  * Registro en archivos de salida
  * Multithreading para acelerar escaneos


````python
import socket

def scan_ports(host, start_port=20, end_port=1024):
    print(f"\nEscaneando puertos del host: {host} (de {start_port} a {end_port})\n")

    try:
        for port in range(start_port, end_port + 1):
            s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
            s.settimeout(0.5)
            result = s.connect_ex((host, port))
            if result == 0:
                print(f"Puerto {port}: ABIERTO")
            s.close()
    except socket.gaierror:
        print("Error: El host no fue encontrado.")
    except socket.error:
        print("Error: No se pudo establecer conexión.")
    
    print("\nEscaneo finalizado.")

if __name__ == "__main__":
    print("Escáner de Puertos Básico con Python")
    target_host = input("Ingresa la dirección IP o dominio a escanear: ")
    scan_ports(target_host)
````
